# Audience Poll: Spatial Data & Python Experience

This interactive form allows you to collect audience familiarity levels with spatial data and Python programming.

**How to use:**
1. Run all cells in order
2. Use the interactive form to record responses for three experience levels
3. Click "Add Response" after entering each count
4. Click "Generate Chart" when done to visualize the results

```python
# Import required libraries
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import numpy as np

print("✓ Libraries loaded successfully")
```

## Interactive Polling Form

Use the form below to record audience responses for three familiarity levels:

- **Beginner**: Little or no experience with spatial data and Python
- **Intermediate**: Some experience with spatial data or Python, comfortable with basic concepts
- **Advanced**: Extensive experience with spatial data and Python programming

```python
# Initialize data storage
poll_data = {
    'Beginner': 0,
    'Intermediate': 0,
    'Advanced': 0
}

# Create widgets for the form
level_dropdown = widgets.Dropdown(
    options=['Beginner', 'Intermediate', 'Advanced'],
    value='Beginner',
    description='Level:',
    style={'description_width': 'initial'}
)

count_input = widgets.IntText(
    value=0,
    description='Count:',
    style={'description_width': 'initial'}
)

add_button = widgets.Button(
    description='Add Response',
    button_style='success',
    icon='plus'
)

reset_button = widgets.Button(
    description='Reset All',
    button_style='danger',
    icon='trash'
)

chart_button = widgets.Button(
    description='Generate Chart',
    button_style='primary',
    icon='bar-chart'
)

output_area = widgets.Output()
status_area = widgets.Output()

# Button click handlers
def on_add_click(b):
    """Add count to selected experience level"""
    level = level_dropdown.value
    count = count_input.value
    poll_data[level] += count
    
    with status_area:
        clear_output(wait=True)
        print(f"✓ Added {count} to {level}")
        print(f"\nCurrent Totals:")
        for key, val in poll_data.items():
            print(f"  {key}: {val}")
        total = sum(poll_data.values())
        print(f"\nTotal Responses: {total}")
    
    # Reset count input
    count_input.value = 0

def on_reset_click(b):
    """Reset all counts to zero"""
    for key in poll_data:
        poll_data[key] = 0
    
    with status_area:
        clear_output(wait=True)
        print("✓ All data reset to zero")
    
    with output_area:
        clear_output()

def on_chart_click(b):
    """Generate visualization of poll results"""
    with output_area:
        clear_output(wait=True)
        
        total = sum(poll_data.values())
        if total == 0:
            print("⚠️ No data to display. Please add responses first.")
            return
        
        # Create figure with two subplots
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
        
        levels = list(poll_data.keys())
        counts = list(poll_data.values())
        colors = ['#e74c3c', '#f39c12', '#27ae60']  # Red, Orange, Green
        
        # Bar chart
        bars = ax1.bar(levels, counts, color=colors, edgecolor='black', linewidth=1.5)
        ax1.set_ylabel('Number of Respondents', fontsize=12, fontweight='bold')
        ax1.set_xlabel('Experience Level', fontsize=12, fontweight='bold')
        ax1.set_title('Audience Experience with Spatial Data & Python', fontsize=14, fontweight='bold', pad=20)
        ax1.grid(axis='y', alpha=0.3, linestyle='--')
        
        # Add count labels on bars
        for bar in bars:
            height = bar.get_height()
            ax1.text(bar.get_x() + bar.get_width()/2., height,
                    f'{int(height)}',
                    ha='center', va='bottom', fontsize=14, fontweight='bold')
        
        # Pie chart
        percentages = [count/total*100 for count in counts]
        wedges, texts, autotexts = ax2.pie(
            counts, 
            labels=levels, 
            colors=colors,
            autopct='%1.1f%%',
            startangle=90,
            explode=(0.05, 0.05, 0.05),
            textprops={'fontsize': 12, 'fontweight': 'bold'}
        )
        
        # Make percentage text more visible
        for autotext in autotexts:
            autotext.set_color('white')
            autotext.set_fontsize(14)
        
        ax2.set_title('Distribution by Experience Level', fontsize=14, fontweight='bold', pad=20)
        
        plt.tight_layout()
        plt.show()
        
        # Print summary statistics
        print(f"\n📊 Summary Statistics:")
        print(f"   Total Respondents: {total}")
        print(f"\n   Breakdown:")
        for level, count in poll_data.items():
            pct = (count/total*100) if total > 0 else 0
            print(f"   • {level}: {count} ({pct:.1f}%)")

# Attach handlers to buttons
add_button.on_click(on_add_click)
reset_button.on_click(on_reset_click)
chart_button.on_click(on_chart_click)

# Layout the form
form_layout = widgets.VBox([
    widgets.HTML("<h3 style='margin-bottom: 10px;'>🎯 Poll Entry Form</h3>"),
    widgets.HBox([level_dropdown, count_input]),
    widgets.HBox([add_button, reset_button, chart_button]),
    status_area,
    widgets.HTML("<hr>"),
    widgets.HTML("<h3>📈 Results Visualization</h3>"),
    output_area
])

display(form_layout)

# Display initial status
with status_area:
    print("Ready to record responses. Select a level, enter count, and click 'Add Response'.")
```

## How This Works

This interactive form demonstrates several key Python concepts:

1. **ipywidgets**: Creates interactive UI elements (dropdowns, buttons, text inputs)
2. **Event handlers**: Functions that respond to button clicks
3. **Data visualization**: matplotlib for creating bar and pie charts
4. **State management**: Storing poll data in a Python dictionary

### Workflow:
1. Select experience level from dropdown
2. Enter count (number of people at that level)
3. Click "Add Response" to record
4. Repeat for all three levels as needed
5. Click "Generate Chart" to visualize results
6. Use "Reset All" to clear data and start over

### Tips:
- You can add multiple entries for the same level (they accumulate)
- The status area shows running totals after each addition
- Charts update each time you click "Generate Chart"
- The pie chart shows percentage distribution automatically
